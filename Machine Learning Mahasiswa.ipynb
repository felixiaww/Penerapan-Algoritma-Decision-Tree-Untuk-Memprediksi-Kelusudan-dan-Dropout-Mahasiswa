```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
# Load dataset
df = pd.read_csv("Predict Students' Dropout and Academic Success.csv")
# Definisikan numeric_features terlebih dahulu
numeric_features = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
# Jalankan perulangan boxplot
for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature}')
    plt.show()
```

```python
# Contoh sederhana untuk mengidentifikasi outliers menggunakan IQR
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1
```

```python
# Filter dataframe untuk hanya menyimpan baris yang tidak mengandung outliers pada kolom numerik
condition = ~((df[numeric_features] < (Q1 - 1.5 * IQR)) | (df[numeric_features] > (Q3 + 1.5 * IQR))).any(axis=1)
df_filtered_numeric = df.loc[condition, numeric_features]

# Menggabungkan kembali dengan kolom kategorikal
# Menggunakan 'string' atau eksplisit 'object' dan 'string'
categorical_features = df.select_dtypes(include=['object', 'string']).columns
df = pd.concat([df_filtered_numeric, df.loc[condition, categorical_features]], axis=1)
```

```python
for feature in numeric_features:
    plt.figure(figsize=(10, 6))
    sns.boxplot(x=df[feature])
    plt.title(f'Box Plot of {feature} (Setelah Dihapus Outlier)')
    plt.show()
```

```python
import sklearn
```

```python
# Memisahkan fitur (X) dan target (y)
X = df_lencoder.drop(columns=['Target'])
y = df_lencoder['Target']
```

```python
from sklearn.model_selection import train_test_split

# membagi dataset menjadi training dan testing
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=1
)
```

```python
# menghitung panjang/jumlah data
print("Jumlah data: ", len(X))
# menghitung panjang/jumlah data pada x_train
print("Jumlah data latih: ", len(x_train))
# menghitung panjang/jumlah data pada x_test
print("Jumlah data test: ", len(x_test))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Inisialisasi model Decision Tree
# (Kita gunakan random_state agar hasil pembagian pohonnya konsisten setiap kali di-run)
model_dt = DecisionTreeClassifier(random_state=42)

# 2. Melatih model menggunakan data training (x_train dan y_train)
model_dt.fit(x_train, y_train)

# 3. Melakukan prediksi pada data testing (x_test)
y_pred = model_dt.predict(x_test)

# 4. Evaluasi Model (Akurasi & Classification Report)
akurasi = accuracy_score(y_test, y_pred)
print(f"Akurasi Model Decision Tree: {akurasi * 100:.2f}%\n")

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# 1. Menampilkan Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

# 2. Visualisasi Confusion Matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model_dt.classes_)
fig, ax = plt.subplots(figsize=(8, 6))
disp.plot(cmap=plt.cm.Blues, ax=ax, values_format='d')

plt.title("Confusion Matrix - Decision Tree Model")
plt.xlabel("Prediksi Model")
plt.ylabel("Data Aktual (Sebenarnya)")
plt.show()
```
